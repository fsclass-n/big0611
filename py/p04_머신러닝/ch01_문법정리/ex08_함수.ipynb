```python
# 파이썬 함수의 구조
# 함수 정의
"""
def 함수_이름([매개변수]):
    수행할_문장
    ...
    [return 반환값]
"""

# 함수 호출
# 함수_이름([인수])
```

```python
# 함수의 이름은 add이고 입력으로 2개의 값을 받으며 반환값(출력값)은 2개의 입력값을 더한 값이다.
def add(a, b): 
    return a + b
```

```python
a = 3
b = 4
c = add(a, b)
print(c)
```

```python
# 매개변수(parameter)와 인수(arguments)
# 매개변수는 함수에 입력으로 전달된 값을 받는 변수, 인수는 함수를 호출할 때 전달하는 입력값을 의미한다.

def add(a, b):  # a, b는 매개변수
    return a+b

print(add(3, 4))  # 3, 4는 인수
```

```python
# 함수의 형태
# : 입력값과 반환값의 존재 유무에 따라 4가지 유형
# 입력값x, 반환값x
# 입력값, 반환값x
# 입력값x, 반환값
# 입력값, 반환값
```

```python
# 1. 일반적인 함수
def add(a, b): 
    result = a + b 
    return result

# 반환값을_받을_변수 = 함수_이름(입력_인수1, 입력_인수2, ...)
a = add(3, 4)
print(a)
```

```python
# 2. 입력값이 없는 함수
def say(): 
    return 'Hi'

# 반환값을_받을_변수 = 함수_이름()
a = say()
print(a) 
```

```python
# 반환값이 없는 함수
def add(a, b): 
    print(f"{a}, {b}의 합은 {a+b}입니다.")

# 함수_이름(입력_인수1, 입력_인수2, ...)
add(3, 4)
```

```python
a = add(3, 4)
print(a)
# None은 값이 '없음'을 나타내는 객체이다.
```

```python
# 입력값도, 반환값도 없는 함수
def say(): 
    print('Hi')

# 함수_이름()
say()
```

```python
# 매개변수를 지정하여 호출하기
def sub(a, b):
    return a - b

# result = sub(7, 3)
result = sub(a=7, b=3)  # a에 7, b에 3을 전달
print(result)
```

```python
result = sub(b=5, a=3)  # b에 5, a에 3을 전달
print(result)
```
