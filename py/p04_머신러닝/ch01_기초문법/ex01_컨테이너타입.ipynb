## 1. 기본 데이터 타입
1. 숫자: 정수(int), 실수(float)
2. 문자열(str): 'True', "10", '''문장''', """문장"""
3. 불(bool): True, False

## 2. 컨터이너 타입
1. 리스트(list): [1, 2, 3, 2, 3], 순서(인덱스), 수정(추가, 삭제)
2. 튜플(tuple): (1, 2, 3, 2, 3), 순서(인덱스), 수정x
3. 세트(set): {1, 2, 3}, 순서(인덱스x)x, 수정, 중복x
4. 딕셔너리(dict): {'키':값, '2':3, '3':3}, 수정, 키는 중복x
